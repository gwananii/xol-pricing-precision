```python
import os, glob, shutil, stat, zipfile, subprocess, sys

home = os.path.expanduser("~")
os.chdir(home)                      # step out of any folder we will delete

# 1. The newest fresh-start zip (kept)
zips = []
for d in (os.path.join(home, "Downloads"), home):
    zips += glob.glob(os.path.join(d, "xol_project_fresh_start*.zip"))
if not zips:
    raise FileNotFoundError("xol_project_fresh_start.zip not found in "
                            "Downloads or " + home)
zip_path = max(zips, key=os.path.getmtime)

# 2. Everything old from this project in your home folder
old = [p for p in glob.glob(os.path.join(home, "xol_project*"))
       if os.path.isdir(p)]
old += [os.path.join(home, d) for d in ("xol_data", "xol_out",
        "paper_figures", "xol_figures") if os.path.isdir(os.path.join(home, d))]
old += [os.path.join(home, f) for f in ("xol_tables.py", "xol_figures.py",
        "make_paper_figure_1.py", "make_paper_figures_4_5.py",
        "make_paper_figures_2_3_S1.py", "check_files.py", "RUN_EVERYTHING.py",
        "xol_finish.py") if os.path.isfile(os.path.join(home, f))]

print("Zip to use (kept):", zip_path)
print("\nWill be DELETED permanently:")
for p in old:
    print("   ", p)
if not old:
    print("    (nothing old found)")

def _force(func, path, _):          # removes read-only files too
    os.chmod(path, stat.S_IWRITE)
    func(path)

if input("\nType YES to delete these and start afresh: ").strip() == "YES":
    for p in old:
        if os.path.isdir(p):
            shutil.rmtree(p, onerror=_force)
        else:
            os.remove(p)
    print("Deleted.")

    # 3. Unzip a clean project and run the test
    with zipfile.ZipFile(zip_path) as z:
        z.extractall(home)
    project = os.path.join(home, "xol_project")
    os.chdir(project)
    print("Fresh project in:", project, "\n")
    p = subprocess.Popen([sys.executable, "RUN_EVERYTHING.py", "--quick"],
                         stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                         text=True, encoding="utf8", errors="replace")
    for line in p.stdout:
        print(line, end="")
    p.wait()
    print("\nExit code:", p.returncode)
else:
    print("Nothing deleted.")
```
